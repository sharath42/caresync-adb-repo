```python
dbutils.widgets.text('system_source','')
dbutils.widgets.text('table_name','')
dbutils.widgets.text('landing_timestamp','')

system_source=dbutils.widgets.get('system_source')
table_name=dbutils.widgets.get('table_name')
landing_timestamp=dbutils.widgets.get('landing_timestamp')
```

```python
from pyspark.sql.functions import *

df=spark.read.parquet(f'/Volumes/ads_caresync/landing/landing_volume/{system_source}/{table_name}/{landing_timestamp}')
```

```python
landing_df=df.withColumn('landing_timestamp',lit(f'{landing_timestamp}'))\
    .withColumn('insert_timestamp',current_timestamp())

landing_df.write\
    .mode('append')\
    .saveAsTable(f'ads_caresync.bronze.{table_name}')
```

```python
%sql
create schema if not exists `ads_caresync`.bronze;
create schema if not exists `ads_caresync`.silver;
```
